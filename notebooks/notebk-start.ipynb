```python
def main():
    name = input("Please enter your name: ")
    age = int(input("Please enter your age: "))
    current_year = 2025
    retirement_age = 67
    years_to_retirement = retirement_age - age
    retirement_year = current_year + years_to_retirement

    if years_to_retirement > 0:
        print(f"Hello, {name}. You can retire in the year {retirement_year} at the age of {retirement_age}.")
    else:
        print(f"Hello, {name}. You are already eligible for retirement.")

if __name__ == "__main__":
    main()
```
